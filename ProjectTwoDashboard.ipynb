```python
# Setup the Jupyter version of Dash
from jupyter_dash import JupyterDash

# Configure the necessary Python module imports for dashboard components
import dash_leaflet as dl
from dash import dcc, html
import plotly.express as px
from dash import dash_table
from dash.dependencies import Input, Output, State
import base64
JupyterDash.infer_jupyter_proxy_config()

# Configure OS routines
import os

# Configure the plotting routines
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


#### FIX ME #####
# change animal_shelter and AnimalShelter to match your CRUD Python module file name and class name
from Project1_Code import AnimalShelter

###########################
# Data Manipulation / Model
###########################
# FIX ME update with your username and password and CRUD Python module name

username = "aacuser"
password = "SNHU1293"

# Connect to database via CRUD Module
shelter = AnimalShelter()

# class read method must support return of list object and accept projection json input
# sending the read method an empty document requests all documents be returned
df = pd.DataFrame.from_records(shelter.read({}))

# MongoDB v5+ is going to return the '_id' column and that is going to have an 
# invlaid object type of 'ObjectID' - which will cause the data_table to crash - so we remove
# it in the dataframe here. The df.drop command allows us to drop the column. If we do not set
# inplace=True - it will reeturn a new dataframe that does not contain the dropped column(s)
#df.drop(columns=['_id'],inplace=True)

## Debug
# print(len(df.to_dict(orient='records')))
# print(df.columns)


#########################
# Dashboard Layout / View
#########################
app = JupyterDash(__name__)

#FIX ME Add in Grazioso Salvare’s logo
image_filename = 'gs_logo.png' # replace with your own image
encoded_image = base64.b64encode(open(image_filename, 'rb').read())

#FIX ME Place the HTML image tag in the line below into the app.layout code according to your design
#FIX ME Also remember to include a unique identifier such as your name or date


app.layout = html.Div([
#    html.Div(id='hidden-div', style={'display':'none'}),
    html.Center(html.B(html.H1('CS-340 Dashboard'))),
    html.Hr(),
    html.Img(src='data:image/png;base64,{}'.format(encoded_image.decode())),
    html.Header("Developed by Alisha Brayboy"),
    html.Div(
        
#FIXME Add in code for the interactive filtering options. For example, Radio buttons, drop down, checkboxes, etc.
    dcc.RadioItems(
      id='filter-type',
      options=[
          {'label': 'Water Rescue', 'value': 'water'},
          {'label': 'Mountain or Wilderness Rescue', 'value': 'mountain'},
          {'label': 'Disaster or Individual Tracking', 'value': 'disaster'},
          {'label': 'Reset', 'value': 'all'},
      ],
value='all', #Default selection
labelStyle={'display': 'inline-block'} # Arrange radio items horizontally
  )
        
    ),
    html.Hr(),
    dash_table.DataTable(id='datatable-id',
                         columns=[{"name": i, "id": i, "deletable": False, "selectable": True} for i in df.columns],
                         data=df.to_dict('records'),
                         
#FIXME: Set up the features for your interactive data table to make it user-friendly for your client
#If you completed the Module Six Assignment, you can copy in the code you created here 
        row_selectable='single',
        page_action='native',
        page_size=10,
        style_table={'height': '300px', 'overflowY': 'auto'},
        style_header={ 'border': '3px solid black', 'backgroundColor': 'black', 'color': 'white', 'fontWeight': 'bold'},
        style_cell={ 'border': '3px solid grey', 'textAlign': 'left', 'padding': '5px' },
        style_data={'backgroundColor': 'rgb(210, 210, 210)','color': 'black'},
                        ),
    html.Br(),
    html.Hr(),
#This sets up the dashboard so that your chart and your geolocation chart are side-by-side
    html.Div(className='row',
         style={'display' : 'flex'},
             children=[
        html.Div(
            id='graph-id',
            className='col s12 m6',

            ),
        html.Div(
            id='map-id',
            className='col s12 m6',
            )
        ])
])

#############################################
# Interaction Between Components / Controller
#############################################

#This callback will update the data table based on the filter selection and return the filtered data and columns to the datatable component
@app.callback(
    [Output('datatable-id', 'data'), Output('datatable-id', 'columns')],
    [Input('filter-type', 'value')]
)
def update_dashboard(filter_type):
    shelter.collection.create_index("animal_type")
    shelter.collection.create_index("breed")
    shelter.collection.create_index("sex_upon_outcome")
    shelter.collection.create_index("age_upon_outcome_in_weeks")
    shelter.collection.create_index([("animal_type", 1), ("breed", 1), ("sex_upon_outcome", 1)])

    pipeline = []

    if filter_type == 'water':
        pipeline = [
            {"$match": {"animal_type": "Dog"}},
            {"$match": {"breed": {"$in": ["Labrador Retriever","Chesapeake Bay Retriever", "Newfoundland"]}}},
            {"$match": {"sex_upon_outcome": "Intact Female"}},
            {"$match": {"age_upon_outcome_in_weeks": {"$gte": 26.0, "$lte": 156.0}}},
            {"$group": {"_id": "$breed", "count": {"$sum": 1}}},
            {"$sort": {"count": -1}}
        ]
    elif filter_type == 'mountain':
        pipeline = [
            {"$match": {"animal_type": "Dog"}},
            {"$match": {"breed": {"$in": ["German Shepherd", "Alaskan Malamute", "Old English Sheepdog", "Siberian Husky", "Rottweiler"]}}},
            {"$match": {"sex_upon_outcome": "Intact Male"}},
            {"$match": {"age_upon_outcome_in_weeks": {"$gte": 26.0, "$lte": 156.0}}}
        ]
    elif filter_type == 'disaster':
        pipeline = [
            {"$match": {"animal_type": "Dog"}},
            {"$match": {"breed": {"$in": ["Doberman Pinscher", "German Shepherd", "Golden Retriever", "Bloodhound", "Rottweiler"]}}},
            {"$match": {"sex_upon_outcome": "Intact Male"}},
            {"$match": {"age_upon_outcome_in_weeks": {"$gte": 20.0, "$lte": 300.0}}}
        ]
    elif filter_type == 'all':
        df = pd.DataFrame.from_records(shelter.read({}))
        columns = [{"name": i, "id": i, "deletable": False, "selectable": True} for i in df.columns]
        data = df.to_dict('records')
        return data, columns

    filtered_data = shelter.collection.aggregate(pipeline)
    df = pd.DataFrame.from_records(filtered_data)
    columns = [{"name": i, "id": i, "deletable": False, "selectable": True} for i in df.columns]
    data = df.to_dict('records')
    return data, columns


#This callback will update the pie chart based on the filtered data from the datatable, it will also update based on the selected row in the datatable
@app.callback(
    Output('graph-id', "children"),
    [Input('datatable-id', "derived_virtual_data")])
def update_graphs(viewData):
    df = pd.DataFrame.from_dict(viewData)
    if len(df) == 0:
        return [dcc.Graph(figure={})]
    else:
        return [
            dcc.Graph(            
                figure = px.pie(df, names='breed', title='Preferred Animals')
            )    
        ]
    

#This call back provides 2 statistical summaries of the data in the datatable - one summary is the count of animals by type and the other is the count of animals by age in weeks. The statistical summaries will update based on the filtered data in the datatable
@app.callback(
    Output('graph-id', "children"),
    [Input('datatable-id', "derived_virtual_data")])
def update_stats(viewData):
    df = pd.DataFrame.from_dict(viewData)
    if len(df) == 0:
        return [html.Div("No data available")]
    else:
        pipeline_by_type = [
            {"$match": {"animal_type": {"$exists": True, "$ne": None}}},
            {"$group": {"_id": "$animal_type", "total_rescued": {"$sum": 1}}},
            {"$sort": {"total_rescued": -1}}
        ]

        pipeline_by_age = [
            {"$match": {"age_upon_outcome_in_weeks": {"$exists": True, "$ne": None}}},
            {"$group": {"_id": "$age_upon_outcome_in_weeks", "count": {"$sum": 1}}},
            {"$sort": {"_id": 1}}
        ]

        type_stats = pd.DataFrame.from_records(shelter.collection.aggregate(pipeline_by_type))
        age_stats = pd.DataFrame.from_records(shelter.collection.aggregate(pipeline_by_age))

        type_stats = type_stats.rename(columns={"_id": "animal_type"})
        age_stats = age_stats.rename(columns={"_id": "age_upon_outcome_in_weeks"})

        return [
            dcc.Graph(
                figure=px.bar(type_stats, x='animal_type', y='total_rescued', title='Rescues by Animal Type')
            ),
            dcc.Graph(
                figure=px.line(age_stats, x='age_upon_outcome_in_weeks', y='count', title='Rescues by Age (weeks)')
            )
        ]

#This callback will highlight a cell on the data table when the user selects it
@app.callback(
    Output('datatable-id', 'style_data_conditional'),
    [Input('datatable-id', 'selected_columns')]
)
def update_styles(selected_columns):
    return [{
        'if': { 'column_id': i },
        'background_color': '#D2F3FF'
    } for i in selected_columns]


# This callback will update the geo-location chart for the selected data entry
# derived_virtual_data will be the set of data available from the datatable in the form of 
# a dictionary.
# derived_virtual_selected_rows will be the selected row(s) in the table in the form of
# a list. For this application, we are only permitting single row selection so there is only
# one value in the list.
# The iloc method allows for a row, column notation to pull data from the datatable
@app.callback(
    Output('map-id', "children"),
    [Input('datatable-id', "derived_virtual_data"),
     Input('datatable-id', "derived_virtual_selected_rows")])
def update_map(viewData, index):  
    if viewData is None:
        return
    elif index is None:
        return
    
    dff = pd.DataFrame.from_dict(viewData)
    # Because we only allow single row selection, the list can be converted to a row index here
    if index is None:
        row = 0
    else: 
        row = index[0]
        
    # Austin TX is at [30.75,-97.48]
    return [
        dl.Map(style={'width': '800px', 'height': '500px'}, center=[30.75,-97.48], zoom=10, children=[
            dl.TileLayer(id="base-layer-id"),
            # Marker with tool tip and popup
            # Column 13 and 14 define the grid-coordinates for the map
            # Column 4 defines the breed for the animal
            # Column 9 defines the name of the animal
            dl.Marker(position=[dff.iloc[row,13],dff.iloc[row,14]], children=[
                dl.Tooltip(dff.iloc[row,4]),
                dl.Popup([
                    html.H1("Animal Name"),
                    html.P(dff.iloc[row,9])
                ])
            ])
        ])
    ]


# Run app and display result in jupyterlab mode, note, if you have previously run a prior app, the default port of 8050 may not be available, if so, try setting an alternate port.
app.run_server() 
```
